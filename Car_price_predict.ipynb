```python
import pandas as pd

# Load dataset
df = pd.read_csv("Quikr car price prediction.csv")

# Convert year to numeric
df["year"] = pd.to_numeric(df["year"], errors="coerce")

# Remove commas and convert Price to numeric
df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace(",", "", regex=False)
)

df["Price"] = pd.to_numeric(df["Price"], errors="coerce")

# Remove "kms" and commas from kms_driven
df["kms_driven"] = (
    df["kms_driven"]
    .astype(str)
    .str.replace(" kms", "", regex=False)
    .str.replace(",", "", regex=False)
)

df["kms_driven"] = pd.to_numeric(df["kms_driven"], errors="coerce")

# Remove rows containing null values
df = df.dropna(subset=["year", "Price", "kms_driven"])

# Convert to integer
df["year"] = df["year"].astype(int)
df["Price"] = df["Price"].astype(int)
df["kms_driven"] = df["kms_driven"].astype(int)

print(df[["year", "Price", "kms_driven"]].head())
print(df[["year", "Price", "kms_driven"]].dtypes)
```
